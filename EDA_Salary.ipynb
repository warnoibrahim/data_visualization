```python
import matplotlib.pyplot as plt

subset = df.groupby('education')['fnlwgt'].sum().reset_index()
subset = subset.sort_values('fnlwgt')

fig, ax = plt.subplots(figsize=(10, 7))

ax.barh(
    subset['education'],
    subset['fnlwgt'] / 1_000_000,
    color='purple'
)

ax.set_title('Total Weight by Education', fontsize=20, fontweight='bold')
ax.set_xlabel('Sum of fnlwgt (Millions)')
ax.set_ylabel('Education')

ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

subset = df.groupby('education')['fnlwgt'].sum().reset_index()

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    subset['education'],
    subset['fnlwgt'] / 1_000_000,
    color='purple',
    marker='o',
    linewidth=2
)

ax.set_title('Total Weight by Education', fontsize=20, fontweight='bold')
ax.set_xlabel('Education')
ax.set_ylabel('Sum of fnlwgt (Millions)')

plt.xticks(rotation=45, ha='right')
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

subset = df.groupby('education')['age'].median().reset_index()

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    subset['education'],
    subset['age'],
    color='purple',
    marker='o',
    linewidth=2
)

ax.set_title('Median Age by Education', fontsize=20, fontweight='bold')
ax.set_xlabel('Education')
ax.set_ylabel('Median Age (Years)')

plt.xticks(rotation=45, ha='right')
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
