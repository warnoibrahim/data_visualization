```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='-.')
plt.show()
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='-.', marker='*', markersize='20')
plt.show()
```

```python
import matplotlib.pyplot as plt

years = [1970, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025, 2030, 2035]
pops = [3.6, 4.0, 4.4, 4.8, 5.3, 5.7, 6.1, 6.5, 7.3, 7.3, 7.3, 7.4, 7.5]

fig, ax = plt.subplots(figsize=(11, 6))
ax.set_facecolor('#F4F7FB')

ax.plot(
    years, pops,
    color='#6366F1',
    linewidth=3,
    marker='o',
    markersize=8,
    markerfacecolor='#FBBF24',
    markeredgecolor='white',
    markeredgewidth=2
)

ax.fill_between(years, pops, color='#6366F1', alpha=0.12)

ax.set_title(
    'World Population Through the Years',
    fontsize=22,
    fontweight='bold',
    color='darkblue',
    pad=20
)
ax.set_xlabel('Year', fontsize=12, color='green')
ax.set_ylabel('Population (Billions)', fontsize=12, color='red')

ax.set_ylim(0, 9)
ax.grid(axis='y', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

years = [1970, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025, 2030, 2035]
pops = [3.6, 4.0, 4.4, 4.8, 5.3, 5.7, 6.1, 6.5, 7.3, 7.3, 7.3, 7.4, 7.5]

fig, ax = plt.subplots(figsize=(11, 6))
ax.set_facecolor('#F4F7FB')

# Draw the line
ax.plot(years, pops, color='#6366F1', linewidth=3)

# Add person symbols at each data point
for year, pop in zip(years, pops):
    ax.text(
        year, pop,
        '👤',
        fontfamily='Segoe UI Emoji',
        fontsize=18,
        ha='center',
        va='center'
    )

# Shade the area below the line
ax.fill_between(years, pops, color='#6366F1', alpha=0.12)

# Add title and axis labels
ax.set_title(
    'World Population Through the Years',
    fontsize=22,
    fontweight='bold',
    color='darkblue',
    pad=20
)
ax.set_xlabel('Year', fontsize=12, color='green')
ax.set_ylabel('Population (Billions)', fontsize=12, color='red')

ax.set_ylim(0, 9)
ax.grid(axis='y', linestyle='--', alpha=0.3)
ax.set_axisbelow(True)

# Remove the top and right borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

years = [1970, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025, 2030, 2035]
pops = [3.6, 4.0, 4.4, 4.8, 5.3, 5.7, 6.1, 6.5, 7.3, 7.3, 7.3, 7.4, 7.5]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(years, pops, color='purple')

# Add person symbols at each data point
for year, pop in zip(years, pops):
    ax.text(
    year, pop, '🧍',
    fontfamily='Segoe UI Emoji',
    fontsize=18,
    ha='center',
    va='center'
)

ax.set_title('World Population', color='blue', fontsize=20, fontweight='bold')
ax.set_xlabel('Growth by Year', color='green')
ax.set_ylabel('Population in Billions', color='red')

ax.grid(True, alpha=0.3)
plt.show()
```
