#### Plot Vocabulary:
-   figure (fig) - container, window to put your plots on
         - figsize - size of the figure in inches
         - dpi - dots per inch (resolution)
         - facecolor - background color
         - subplotpars - distance between subplots
-   axes (ax) - area within the fig where data is plotted
         - title - title for the chart/plot
         - xlabel, ylabel, xscale, yscale,, xlim, ylim
-   axis - x, y axis of a plot or chart
-   line - data itself, cant be fully customized
         - color
         - linestyle
         - linewidth
         - markers - symbols that represent data points
         - markersize
-   Legend - small container with labels/descriptors of elements
-   labels - names/labels for the x and y axis
-   title - title for the chart/plot
-   markers - symbols that represent data points
-   grid - backgrod (attr)
-   grid - backgrod (attribute that can be edited)
-   ticks (tic) - marks along the x,y axis to represent values

```python
import matplotlib.pyplot as plt

x = [80, 85, 90, 95, 100, 105, 110, 115, 120, 125]
y = [240, 250, 260, 170, 190, 160, 300, 310, 320, 330]

# Create a lineplot, custom color, dotted linestyle, diamond markers
plt.plot(
    x, y,
    linestyle=':',
    marker='D',
    color=(0.5, 0.5, 0.8)
)
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')

plt.grid()

plt.show()
```

```python
# Create a lineplot, custom color, dotted linestyle, diamond markers
plt.plot(
    x, y,
    linestyle='--',
    marker='D',
    color=(0.5, 0.5, 0.8)
)
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')

plt.grid(color='red', linestyle='--', alpha=0.4)

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [80, 85, 90, 95, 100, 105, 110, 115, 120, 125]
y = [240, 250, 260, 170, 190, 160, 300, 310, 320, 330]

fig, ax = plt.subplots(figsize=(10, 5))

# facecolor: colors the chart background
ax.set_facecolor('lightblue')

# Draw the line with circle markers
ax.plot(x, y, color='purple', marker='o', linestyle='--')

# fill_between: shades the area between the line and zero
ax.fill_between(x, y, 0, color='purple', alpha=0.3)

ax.set_title('Sports Watch Data', fontsize=20, fontweight='bold')
ax.set_xlabel('Average Pulse')
ax.set_ylabel('Calories Burned')

ax.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

x1 = np.array([0, 1, 2, 3])
y1 = np.array([10, 20, 25, 30])

# Create 1 row with 2 columns
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(x, y)
ax[0].set_title('First Graph')

ax[1].plot(x1, y1)
ax[1].set_title('Second Graph')

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x)

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sin Function')

axes[0, 1].plot(x, y2)
axes[0, 1].set_title('Cosine Function')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Tangent Function')

axes[1, 1].plot(x, y4)
axes[1, 1].set_title('Exponential Function')

plt.tight_layout()
plt.show()
```

```python
years=[1950,1955,1960,1965,1970,1980,1985,1990,1995,2000,2005,2010,2015]
pops=[2.5,2.7,3.0,3.3,3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3]
death=[1.2,1.1,1.2,2.1,2.0,2.3,1.8,1.9,2.6,1.6,2.4,2.4,4.0]
job_rate = [0.5, 0.6, 0.8, 0.9, 1.0, 1.2, 1.5, 1.7, 2.0, 2.2, 2.5, 2.7, 2.9]


plt.plot(years, pops, color='red', linestyle='--', label='Population')
plt.plot(years, death, color='green', linestyle=':', label='Death')
plt.plot(years, job_rate, color='purple', linestyle='-.', label='Job Rate')

plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population (Billions)')

plt.legend() # when using a legend, the plots must have labels. See .plot() above
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```

```python
years=[1950,1955,1960,1965,1970,1980,1985,1990,1995,2000,2005,2010,2015]
pops=[2.5,2.7,3.0,3.3,3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3]
death=[1.2,1.1,1.2,2.1,2.0,2.3,1.8,1.9,2.6,1.6,2.4,2.4,4.0]
job_rate = [0.5, 0.6, 0.8, 0.9, 1.0, 1.2, 1.5, 1.7, 2.0, 2.2, 2.5, 2.7, 2.9]


plt.plot(years, pops, color='red', linestyle='--', label='Population')
plt.plot(years, death, color='green', linestyle=':', label='Death')
plt.plot(years, job_rate, color='purple', linestyle='-.', label='Job Rate')

plt.fill_between(years, pops, alpha=0.3)
plt.fill_between(years, death, alpha=0.3)
plt.fill_between(years, job_rate, alpha=0.3)

plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population (Billions)')

plt.legend()
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```
